```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

### Análise dos Gráficos - Diferença entre Meios de Transporte por Raça

- **Carro:**
  - Mulheres Brancas: X%
  - Mulheres Pretas ou Pardas: Y%
  - *Diferença:* Mulheres Brancas - Mulheres Pretas ou Pardas = Z%

- **Transporte Coletivo:**
  - Mulheres Pretas ou Pardas: X%
  - Mulheres Indígenas: Y%
  - *Diferença:* Mulheres Pretas ou Pardas - Mulheres Indígenas = Z%

- **Caminhada / A pé:**
  - Mulheres Indígenas: X%
  - Mulheres Brancas: Y%
  - *Diferença:* Mulheres Indígenas - Mulheres Brancas = Z%

```python
# 1. Leitura da aba BR GR UF MU
df_uf = pd.read_excel("seu_arquivo.xlsx", sheet_name="BR GR UF MU")

# 2. Listar apenas os estados únicos
estados = df_uf["UF"].dropna().unique()
print("Lista de Estados:")
print(sorted(estados))

# Transformar as colunas de meios de transporte em linhas com melt
df_melt = pd.melt(
    df_uf,
    id_vars=["UF", "Município"],
    value_vars=["Transporte Coletivo", "Carro", "Bicicleta", "A pé"],
    var_name="Meio de Transporte",
    value_name="Quantidade",
)

# Resposta 1: Estado com mais mulheres em transporte coletivo
est_mais_coletivo = (
    df_melt[df_melt["Meio de Transporte"] == "Transporte Coletivo"]
    .groupby("UF")["Quantidade"]
    .sum()
    .idxmax()
)

# Resposta 2: Estado com menos mulheres em transporte coletivo
est_menos_coletivo = (
    df_melt[df_melt["Meio de Transporte"] == "Transporte Coletivo"]
    .groupby("UF")["Quantidade"]
    .sum()
    .idxmin()
)

# Resposta 3: Cidade com mais mulheres andando de bicicleta
cid_mais_bike = (
    df_melt[df_melt["Meio de Transporte"] == "Bicicleta"]
    .groupby(["Município", "UF"])["Quantidade"]
    .sum()
    .idxmax()
)

# Resposta 4: Cidade com mais mulheres utilizando carro
cid_mais_carro = (
    df_melt[df_melt["Meio de Transporte"] == "Carro"]
    .groupby(["Município", "UF"])["Quantidade"]
    .sum()
    .idxmax()
)
```

```python
# 1. Leitura do arquivo (aba Exemplo Gráfico Brasil)
df = pd.read_excel("seu_arquivo.xlsx", sheet_name="Exemplo Gráfico Brasil")

# 2. Configuração do eixo e cores
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
cores = ["#ff9999", "#66b3ff", "#99ff99", "#ffcc99", "#c2c2f0", "#ffb3e6"]
grupos = ["Branca", "Preta ou Parda", "Indígena"]

# 3. Iteração para gerar um gráfico por grupo de raça/cor
for i, col in enumerate(grupos):
    axes[i].pie(
        df[col],
        labels=df["Meio de Transporte"],
        autopct="%1.3f%%",  # 3 casas decimais conforme solicitado
        startangle=180,
        colors=cores,
        wedgeprops={"edgecolor": "white", "linewidth": 1.2},
    )
    axes[i].set_title(f"Cor/Raça: {col}", fontsize=12, fontweight="bold")

plt.suptitle(
    "Distribuição de Meios de Transporte por Cor/Raça",
    fontsize=14,
    fontweight="bold",
)
plt.tight_layout()
plt.savefig("grafico_transporte.png", dpi=300)
plt.show()
```
