```python
import numpy as np
import pandas as pd

print("pandas:", pd.__version__)
print("numpy:", np.__version__)
```

```python
caminho = r"H:\Python\3°Etapa\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"

tabela = pd.read_excel(caminho, header = 2)

tabela
```

```python
tabela_arrumada = tabela.iloc[0:5570, 1:]

tabela_arrumada
```

```python
tabela_arrumada["Diferença"] = tabela_arrumada["População 2010 (Alteração de limites até 2022)1"] - tabela_arrumada ["População do Censo 2022"]

tabela_arrumada
```

```python
tabela_arrumada.to_csv(r"H:\Python\3°Etapa\Tabela_Diferença.csv" , sep = ";")
```

```python
tabela_estado =  (
    tabela_arrumada.groupby("UF")[["População de Censo 2022", "População do Município 2010\n(Sinopse)", "Diferença"]]

    .sum()
    .reset_index()
)

tabela_estado
```

```python
tabela_municipio =  (
    tabela_arrumada.groupby("NOME DO MUNICÍPIO")[["População de Censo 2022", "População do Município 2010\n(Sinopse)", "Diferença"]]

    .sum()
    .reset_index()
)

tabela_estado
```
